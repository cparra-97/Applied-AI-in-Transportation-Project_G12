```python
# Load data

import pandas as pd

df_01 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202401.csv')
df_02 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202402.csv')
df_03 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202403.csv')
df_04 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202404.csv')
df_05 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202405.csv')
df_06 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202406.csv')
df_07 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202407.csv')
df_08 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202408.csv')
df_09 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202409.csv')
df_10 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202410.csv')
df_11 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202411.csv')
df_12 = pd.read_csv(r'C:/Users/caral/OneDrive - KTH/SUPD/Year 2/AI in Transportation/202412.csv')

df_all = pd.concat([df_01, df_02, df_03, df_04, df_05, df_06, df_07, df_08, df_09, df_10, df_11, df_12], ignore_index=True)

df_all.head(10)
```

```python
# Analysis

import matplotlib.pyplot as plt
import seaborn as sns

print(df_all['observed_arrival_delay'].mean()+3*df_all['observed_arrival_delay'].std())

df_rain = df_all[df_all['rain']>0]
df_no_rain = df_all[df_all['rain']==0]

df_rain = df_rain.dropna(subset=['observed_arrival_delay'])
df_no_rain = df_no_rain.dropna(subset=['observed_arrival_delay'])

print(df_rain['observed_arrival_delay'].describe())
print(df_no_rain['observed_arrival_delay'].describe())
```

```python
# Datasets for each model

import numpy as np

df_model = df_all.drop(['date','previous_StopAreaID','observed_arrival_time_sfm','LineID_direction',
                        'date_time',
                        # 'x_dest','y_dest', 
                        'date_format'], axis=1)
df_model = df_model.dropna(subset=['observed_arrival_delay'])
df_model = df_model.dropna(subset=['coming_from_stop']) ###### when it is the first stop
df_model = df_model.dropna(subset=['hum']) ############# when there is no weather info

cond = [(df_model['observed_arrival_delay'] < -60),
        (df_model['observed_arrival_delay'] >= -60) & (df_model['observed_arrival_delay'] < 150),
        (df_model['observed_arrival_delay'] >= 150) & (df_model['observed_arrival_delay'] < 420),
        (df_model['observed_arrival_delay'] >= 420)]
res = [0,1,2,3]
df_model['delay_cat'] = np.select(cond, res)

df_model = df_model[df_model['rain']>0] 
df_model = df_model[df_model['observed_arrival_delay']<df_all['observed_arrival_delay'].mean()+3*df_all['observed_arrival_delay'].std()] ########## Removing outlayers >>>> check
df_model = df_model[df_model['observed_arrival_delay']>df_all['observed_arrival_delay'].mean()-3*df_all['observed_arrival_delay'].std()] ######### Removing outlayers >>>> check
# df_model = df_model[df_model['observed_arrival_delay']>0] ######### Removing outlayers >>>> check


x3 = df_model.drop(['observed_arrival_delay','delay_cat'], axis=1) # All variables
x2 = x3.drop(['coming_from_origin','coming_from_stop', 'x_origin', 'y_origin', 'x_prevstop', 'y_prevstop',
             'x_dest', 'y_dest'], axis=1) # Without the spatial information
x1 = x2.drop(['wind_peak','solar_rad','rain','hum','temp','air_pre','wind_spe','wind_dir',
             'wind_peak-1','solar_rad-1','rain-1','hum-1','temp-1','air_pre-1','wind_spe-1','wind_dir-1',
             'wind_peak-2','solar_rad-2','rain-2','hum-2','temp-2','air_pre-2','wind_spe-2','wind_dir-2'], axis=1) # Just SL information
y_val = df_model['observed_arrival_delay'] # For numerical models
y_cat = df_model['delay_cat'] # For classification models
```

```python
# Scaling and train / test split

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Define columns to encode and scale
columns_to_encode = ['LineID', 'observed_StopAreaID', 'month', 'week_day']
columns_to_scale1 = ['stop_sequence', 'dist_traveled', 'scheduled_arrival_time_sfm', 'ini_delay']
columns_to_scale2 = ['stop_sequence', 'dist_traveled', 'scheduled_arrival_time_sfm', 'ini_delay', 
                    'wind_peak', 'solar_rad', 'rain', 'hum', 'temp', 'air_pre', 'wind_spe', 'wind_dir', 
                    'wind_peak-1', 'solar_rad-1', 'rain-1', 'hum-1', 'temp-1', 'air_pre-1', 'wind_spe-1', 'wind_dir-1',
                    'wind_peak-2', 'solar_rad-2', 'rain-2', 'hum-2', 'temp-2', 'air_pre-2', 'wind_spe-2', 'wind_dir-2']
columns_to_scale3 = ['stop_sequence', 'dist_traveled', 'scheduled_arrival_time_sfm', 'ini_delay', 
                    'wind_peak', 'solar_rad', 'rain', 'hum', 'temp', 'air_pre', 'wind_spe', 'wind_dir', 
                    'wind_peak-1', 'solar_rad-1', 'rain-1', 'hum-1', 'temp-1', 'air_pre-1', 'wind_spe-1', 'wind_dir-1',
                    'wind_peak-2', 'solar_rad-2', 'rain-2', 'hum-2', 'temp-2', 'air_pre-2', 'wind_spe-2', 'wind_dir-2', 
                    'x_prevstop', 'y_prevstop', 'x_origin', 'y_origin', 'coming_from_origin', 'coming_from_stop',
                    'x_dest', 'y_dest',]

# Instantiate encoder/scaler
scaler = StandardScaler()
ohe = OneHotEncoder(sparse_output=False)

#Scale and encode

x1_scaled_columns  = scaler.fit_transform(x1[columns_to_scale1])
x1_encoded_columns = ohe.fit_transform(x1[columns_to_encode])
x1_scaled = np.concatenate([x1_scaled_columns, x1_encoded_columns], axis=1)

x2_scaled_columns  = scaler.fit_transform(x2[columns_to_scale2])
x2_encoded_columns = ohe.fit_transform(x2[columns_to_encode])
x2_scaled = np.concatenate([x2_scaled_columns, x2_encoded_columns], axis=1)

x3_scaled_columns  = scaler.fit_transform(x3[columns_to_scale3])
x3_encoded_columns = ohe.fit_transform(x3[columns_to_encode])
x3_scaled = np.concatenate([x3_scaled_columns, x3_encoded_columns], axis=1)

# Set the split

X1_re_train_scaled, X1_re_test_scaled, y1_re_train, y1_re_test = train_test_split(x1_scaled, y_val, test_size=0.2, random_state=42)
X2_re_train_scaled, X2_re_test_scaled, y2_re_train, y2_re_test = train_test_split(x2_scaled, y_val, test_size=0.2, random_state=42)
X3_re_train_scaled, X3_re_test_scaled, y3_re_train, y3_re_test = train_test_split(x3_scaled, y_val, test_size=0.2, random_state=42)

X1_cl_train_scaled, X1_cl_test_scaled, y1_cl_train, y1_cl_test = train_test_split(x1_scaled, y_cat, test_size=0.2, random_state=42)
X2_cl_train_scaled, X2_cl_test_scaled, y2_cl_train, y2_cl_test = train_test_split(x2_scaled, y_cat, test_size=0.2, random_state=42)
X3_cl_train_scaled, X3_cl_test_scaled, y3_cl_train, y3_cl_test = train_test_split(x3_scaled, y_cat, test_size=0.2, random_state=42)
```

```python
# Linear regression

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import seaborn as sns
import matplotlib.pyplot as plt

# Create a Linear Regression model
model_re1 = LinearRegression()
model_re2 = LinearRegression()
model_re3 = LinearRegression()

# Fit the model to the training data
model_re1.fit(X1_re_train_scaled, y1_re_train)
model_re2.fit(X2_re_train_scaled, y2_re_train)
model_re3.fit(X3_re_train_scaled, y3_re_train)

# Predict the test data with the fitted model
y1_re_pred = model_re1.predict(X1_re_test_scaled)
y2_re_pred = model_re2.predict(X2_re_test_scaled)
y3_re_pred = model_re3.predict(X3_re_test_scaled)

mae = mean_absolute_error(y1_re_test, y1_re_pred)
mse = mean_squared_error(y1_re_test, y1_re_pred)
r2 = r2_score(y1_re_test, y1_re_pred)
print('LINEAR REGRESSION MODEL 1')
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
print('')

plt.figure(figsize=(8, 6))
plt.scatter(y1_re_test, y1_re_pred, alpha=0.5)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs. Predicted Values")
plt.plot([min(y1_re_test), max(y1_re_test)], [min(y1_re_test), max(y1_re_test)], linestyle='--', color='red', lw=2)
plt.show()

mae = mean_absolute_error(y2_re_test, y2_re_pred)
mse = mean_squared_error(y2_re_test, y2_re_pred)
r2 = r2_score(y2_re_test, y2_re_pred)
print('LINEAR REGRESSION MODEL 2')
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
print('')

plt.figure(figsize=(8, 6))
plt.scatter(y2_re_test, y2_re_pred, alpha=0.5)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs. Predicted Values")
plt.plot([min(y2_re_test), max(y2_re_test)], [min(y2_re_test), max(y2_re_test)], linestyle='--', color='red', lw=2)
plt.show()

mae = mean_absolute_error(y3_re_test, y3_re_pred)
mse = mean_squared_error(y3_re_test, y3_re_pred)
r2 = r2_score(y3_re_test, y3_re_pred)
print('LINEAR REGRESSION MODEL 3')
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
print('')

plt.figure(figsize=(8, 6))
plt.scatter(y3_re_test, y3_re_pred, alpha=0.5)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs. Predicted Values")
plt.plot([min(y3_re_test), max(y3_re_test)], [min(y3_re_test), max(y3_re_test)], linestyle='--', color='red', lw=2)
plt.show()
```

```python
# XGBoost with Grid Search

from sklearn.model_selection import GridSearchCV
from xgboost import XGBRegressor
import xgboost as xgb

xgb = XGBRegressor(random_state=42, objective='reg:squarederror')

# Define a parameter grid for hyperparameter tuning
param_grid_xgb = {
    'n_estimators': [400, 600, 800],
    'max_depth': [6, 7, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.8]
}

# Finding best parameters for model 1

# Create the GridSearchCV object
grid_xgb = GridSearchCV(xgb, param_grid_xgb, cv=4, scoring='neg_mean_squared_error', verbose=2)

# Fit the grid search to the scaled training data
grid_xgb.fit(X1_re_train_scaled, y1_re_train)

# Get the best parameters
print("Best XGBoost parameters:", grid_xgb.best_params_)
print("Best Score:", grid_xgb.best_score_)

# Create an XGBoost model with the best parameters from the grid search
best_xgb = grid_xgb.best_estimator_
best_xgb.fit(X1_re_train_scaled, y1_re_train)

y_pred = best_xgb.predict(X1_re_test_scaled)
mae = mean_absolute_error(y1_re_test, y_pred)
mse = mean_squared_error(y1_re_test, y_pred)
r2 = r2_score(y1_re_test, y_pred)
print('XGBOOST REGRESSION MODEL 1')
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")

plt.figure(figsize=(8, 6))
plt.scatter(y1_re_test, y_pred, alpha=0.5)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs. Predicted Values")
plt.plot([min(y1_re_test), max(y1_re_test)], [min(y1_re_test), max(y1_re_test)], linestyle='--', color='red', lw=2)
plt.show()


# Finding best parameters for model 2

# Create the GridSearchCV object
grid_xgb = GridSearchCV(xgb, param_grid_xgb, cv=4, scoring='neg_mean_squared_error', verbose=2)

# Fit the grid search to the scaled training data
grid_xgb.fit(X2_re_train_scaled, y2_re_train)

# Get the best parameters
print("Best XGBoost parameters:", grid_xgb.best_params_)
print("Best Score:", grid_xgb.best_score_)

# Create an XGBoost model with the best parameters from the grid search
best_xgb = grid_xgb.best_estimator_
best_xgb.fit(X2_re_train_scaled, y2_re_train)

y_pred = best_xgb.predict(X2_re_test_scaled)
mae = mean_absolute_error(y2_re_test, y_pred)
mse = mean_squared_error(y2_re_test, y_pred)
r2 = r2_score(y2_re_test, y_pred)
print('XGBOOST REGRESSION MODEL 2')
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")

plt.figure(figsize=(8, 6))
plt.scatter(y2_re_test, y_pred, alpha=0.5)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs. Predicted Values")
plt.plot([min(y2_re_test), max(y2_re_test)], [min(y2_re_test), max(y2_re_test)], linestyle='--', color='red', lw=2)
plt.show()


# Finding best parameters for model 3

# Create the GridSearchCV object
grid_xgb = GridSearchCV(xgb, param_grid_xgb, cv=4, scoring='neg_mean_squared_error', verbose=2)

# Fit the grid search to the scaled training data
grid_xgb.fit(X3_re_train_scaled, y3_re_train)

# Get the best parameters
print("Best XGBoost parameters:", grid_xgb.best_params_)
print("Best Score:", grid_xgb.best_score_)

# Create an XGBoost model with the best parameters from the grid search
best_xgb = grid_xgb.best_estimator_
best_xgb.fit(X3_re_train_scaled, y3_re_train)

y_pred = best_xgb.predict(X3_re_test_scaled)
mae = mean_absolute_error(y3_re_test, y_pred)
mse = mean_squared_error(y3_re_test, y_pred)
r2 = r2_score(y3_re_test, y_pred)
print('XGBOOST REGRESSION MODEL 3')
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
print('')

plt.figure(figsize=(8, 6))
plt.scatter(y3_re_test, y_pred, alpha=0.5)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs. Predicted Values")
plt.plot([min(y3_re_test), max(y3_re_test)], [min(y3_re_test), max(y3_re_test)], linestyle='--', color='red', lw=2)
plt.show()
```
