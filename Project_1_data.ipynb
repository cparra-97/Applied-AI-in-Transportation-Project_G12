```python
# Classification of delay

cond = [(df_model['observed_arrival_delay'] < -60),
        (df_model['observed_arrival_delay'] >= -60) & (df_model['observed_arrival_delay'] < 150),
        (df_model['observed_arrival_delay'] >= 150) & (df_model['observed_arrival_delay'] < 480),
        (df_model['observed_arrival_delay'] >= 480)]

res = [0,1,2,3]

df_model['delay_cat'] = np.select(cond, res)
```

```python
# Merge of weather information

import numpy as np

df_model['date_time'] = np.where(df_model['scheduled_arrival_time_sfm']//3600 < 10, 
                                       df_model['date'].astype(str) + '_0' + (df_model['scheduled_arrival_time_sfm']//3600).astype(str), 
                                       df_model['date'].astype(str) + '_' + (df_model['scheduled_arrival_time_sfm']//3600).astype(str))

df_model = pd.merge(df_model, df_wthr, on='date_time', how='left')
df_model.rename(columns={"date_x": "date"}, inplace=True, errors="raise")

df_model.drop(['date_y', 'time'], axis=1, inplace=True)
```

```python
# Defining coordinates and direction

import math

df_routes = pd.merge(df_routes, df_allstop, on='observed_StopAreaID', how='left')
df_routes.rename(columns={"x": "x_origin", "y": "y_origin"}, inplace=True, errors="raise")
df_model = pd.merge(df_model, df_allstop, on='observed_StopAreaID', how='left')
df_model.rename(columns={"x": "x_dest", "y": "y_dest"}, inplace=True, errors="raise")
df_allstop.rename(columns={"observed_StopAreaID": "previous_StopAreaID"}, inplace=True, errors="raise")
df_model = pd.merge(df_model, df_allstop, on='previous_StopAreaID', how='left')
df_allstop.rename(columns={"previous_StopAreaID": "observed_StopAreaID"}, inplace=True, errors="raise")
df_model.rename(columns={"x": "x_prevstop", "y": "y_prevstop"}, inplace=True, errors="raise")
df_model = pd.merge(df_model, df_routes[['LineID_direction', 'x_origin', 'y_origin']], on='LineID_direction', how='left')
df_model['coming_from_origin'] = np.degrees(np.arctan2(df_model['y_origin'] - df_model['y_dest'], df_model['x_origin'] - df_model['x_dest']))
df_model['coming_from_stop'] = np.degrees(np.arctan2(df_model['y_prevstop'] - df_model['y_dest'], df_model['x_prevstop'] - df_model['x_dest']))
```

```python
# Get day of the week and month

# df_model['day'] = df_model['date'].str[-2:].astype(int)
df_model['month'] = df_model['date'].astype(str).str[-4:-2].astype(int)
# df_model['yr'] = df_model['date'].str[:4].astype(int)
df_model['date_format'] = pd.to_datetime(df_model['date'], format='%Y%m%d')
df_model['week_day'] = df_model['date_format'].dt.isocalendar().day.astype(int)
 
```
